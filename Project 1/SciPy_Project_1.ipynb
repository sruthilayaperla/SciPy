```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
# Student data
data = {"Student": ["A", "B", "C", "D", "E", "F", "G", "H","I", "J"],
    "Maths": [78, 85, 67, 90, 56, 72, 88, 95, 64, 81],
    "Physics": [75, 82, 70, 92, 60, 68, 85, 90, 65, 79],
    "Hours": [5, 6, 4, 8, 3, 5, 7, 9, 4, 6]}
df = pd.DataFrame(data)
print(df)
```

```python
# Mean
print("\nMean Maths:")
print(df["Maths"].mean())
```

```python
# Median
print("\nMedian Maths:")
print(df["Maths"].median())
```

```python
# Standard deviation
print("\n standard deviation Maths:")
print(df["Maths"].std())
```

```python
# Variance
print("\nVariance:")
print(df["Maths"].var())
```

```python
# Z-score
z_scores = stats.zscore(df["Maths"])

df["Maths_ZScore"] = z_scores

print("\nZ-Scores:")
print(df)
```

```python
# Correlation
correlation = stats.pearsonr(df["Hours"], df["Maths"])

print("\nCorrelation between Study Hours and Maths:")
print(correlation)
```

```python
# T-test
t_stat, p_value = stats.ttest_1samp(df["Maths"], 75)

print("\nOne Sample T-Test:")
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
# Histogram
plt.hist(df["Maths"], bins=5)
plt.xlabel("Maths Marks")
plt.ylabel("Number of Students")
plt.title("Student Maths Performance")
plt.show()
```
