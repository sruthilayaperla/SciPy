**Optimization** - The optimize module is extremely important.

It can perform:


*   Function Minimization

*   Fucnon Maximization

*   Root Finding
*   Curve Fitting


*   Least Squares


*   Constrained optimization



```python
#Find the minimum of a function
from scipy.optimize import minimize
def f(x):
  return x**2+5*x+6
result=minimize(f,x0=[0])
print(result)
```

```python
#Find the maximum of a function
from scipy.optimize import minimize
def f(x):
    return -x**2 + 5*x + 6
# To maximize f(x), minimize -f(x)
def negative_f(x):
    return -f(x)
result = minimize(negative_f, x0=[0])
print("Maximum value:", f(result.x))
print("At x =", result.x)
```

```python
#Root Finding
from scipy.optimize import root
def equation(x):
  return x**2-4
result=root(equation,1)
print(result.x)
```

**HYPOTHESIS TESTING :**

    Test                and it's SciPy Function
  
*  t-test               and its function is ttest_ind()
*  paired t-test        and its function is ttest_rel()


*   One-sample t-test   and its function is ttest_lsamp()

*   Chi-Square          and its function is chi2_contingency()
*   ANNOVA              and its function is f_oneway()


*   Pearson Corraltion and its function is Pearsonr()

*   Speaman Correlation and its function is Spearmanr()

*   Mann-Whitney and its function is manwhitney()
*   Wilcoxon and its function is wilcoxon()


*   SHapiro-wilk and its function is Shapiro()













```python
#Independent t-test
from scipy.stats import ttest_ind
group1=[10,12,14,15,13]
group2=[20,22,21,23,24]
result=ttest_ind(group1,group2)
print(result)
```

```python
#One-sample ttest
from scipy.stats import ttest_1samp
dtat=[10,12,11,13,14]
result=ttest_1samp(dtat,13)
print(result)
```

```python
#Paired ttest
from scipy.stats import ttest_rel
before=[70,72,75,68,74]
after=[75,78,79,73,80]
result=ttest_rel(before,after)
print(result)
```

```python
#Chi-square test
from scipy.stats import chi2_contingency
data=[[20,30],[25,35]]
chi2,p,dof,expected=chi2_contingency(data)
print("chi-square=",chi2)
print("p-value=",p)
print("Degrees of freedom=",dof)
print("expected=",expected)
```

```python
#ANNOVA : It is used to compare means of multiple groups
from scipy.stats import f_oneway
group1=[10,12,11,13]
group2=[20,22,21,23]
group3=[30,31,29,32]
result=f_oneway(group1,group2,group3)
print(result)
```

```python
#Pearson Correlation:
from scipy.stats import pearsonr
x=[1,2,3,4,5]
y=[2,4,6,8,10]
correlation,pvalue=pearsonr(x,y)
print("correlation=",correlation)
print("p-pvalue=",pvalue)
```

```python
#Spearman Coorelation:
from scipy.stats import spearmanr
x=[1,2,3,4,5]
y=[5,10,15,20,25]
result=spearmanr(x,y)
print(result)
```

```python
#Shapiro-wilk test: Used to test whether data is consistent with a normal distribution
from scipy.stats import shapiro
data=[12,14,15,13,16,17,15]
result=shapiro(data)
print(result)
```

**Signal Filter:** SciPy provides functions such as butter(),filter(),filtfilt()

```python
from scipy.signal import butter
b,a=butter(3,0.2)
print(b)
print(a)
```

```python
import numpy as np
from scipy.signal import butter, lfilter

# 1. Design the filter coefficients (b, a)
b, a = butter(3, 0.2)

# 2. Create a dummy signal 'x' to filter (e.g., a sine wave with some noise)
t = np.linspace(0, 1, 100)
x = np.sin(2 * np.pi * 5 * t) + 0.5 * np.random.randn(100)

# 3. Apply the filter to the signal 'x'
filtered_signal = lfilter(b, a, x)

print("Filter Coefficients b:", b)
print("Filter Coefficients a:", a)
print("First 5 filtered values:", filtered_signal[:5])
```

**Euclidean Distance:** It represents the straight-line distance between two points in a Euclidean space.

**MANHATTAN DISTANCE:** Manhattan Distance is the distance between two points calculated by adding the absolute differences between their coordinates. It is also called City Block Distance because it resembles travelling along city streets in a grid.

```python
#Manhattan Distance
from scipy.spatial.distance import cityblock
a=[1,2]
b=[4,6]
print(cityblock(a,b))
```

**KD Tree:** The KD Tree implementation is used to find the nearest neighbours and distance computation.

```python
from scipy.spatial import KDTree
points=[(1,-1),(2,3),(-2,3),(2,-3)]
kdtree=KDTree(points)
res=kdtree.query((1,1))
print(res)
```
