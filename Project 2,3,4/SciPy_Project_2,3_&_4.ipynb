```python
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt

data = {
    "Employee": ["E1","E2","E3","E4","E5","E6",
                 "E7","E8","E9","E10"],
    "Department": ["IT","IT","HR","HR","Sales",
                   "Sales","IT","HR","Sales","IT"],
    "Experience": [2,5,3,8,4,6,7,10,5,9],
    "Salary": [30000,50000,35000,65000,40000,
               55000,70000,75000,48000,80000]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("\nMean Salary:")
print(df["Salary"].mean())
```

```python
print("\nMedian Salary:")
print(df["Salary"].median())
```

```python
print("\nSalary Standard Deviation:")
print(df["Salary"].std())
```

```python
# Salary Z-score
df["Salary_ZScore"] = stats.zscore(df["Salary"])

print("\nSalary Z-Scores:")
print(df)
```

```python
# Correlation
r, p = stats.pearsonr(df["Experience"], df["Salary"])
print("\nExperience-Salary Correlation")
print("Correlation:", r)
print("P-value:", p)
```

```python
# Department salary
print("\nAverage Salary by Department:")
print(df.groupby("Department")["Salary"].mean())
```

```python
# Boxplot
df.boxplot(column="Salary", by="Department")
plt.title("Salary Distribution by Department")
plt.suptitle("")
plt.ylabel("Salary")
plt.show()
```

#   Project -3 SciPy + Pandas + Matplotlib Data Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

data = {
    "Month": [
        "Jan", "Feb", "Mar", "Apr",
        "May", "Jun", "Jul", "Aug",
        "Sep", "Oct", "Nov", "Dec"
    ],

    "Sales": [
        12000, 15000, 14000, 18000,
        20000, 22000, 25000, 24000,
        28000, 30000, 32000, 35000
    ],

    "Advertising": [
        2000, 2500, 2300, 3000,
        3200, 3500, 4000, 3800,
        4500, 5000, 5200, 5500
    ]
}
df=pd.DataFrame(data)
print(df)
```

```python
# Descriptive statistics
print("\nStatistics:")
print(df.describe())
```

```python
# Correlation
r, p = stats.pearsonr(
    df["Advertising"],
    df["Sales"]
)

print("\nAdvertising vs Sales")
print("Correlation:", r)
print("P-value:", p)
```

```python
 #Z-score
df["Sales_ZScore"] = stats.zscore(
    df["Sales"]
)

print("\nZ-score:")
print(df)
```

```python
# Sales graph
plt.plot(
    df["Month"],
    df["Sales"],
    marker="o"
)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
# Advertising vs Sales
plt.scatter(
    df["Advertising"],
    df["Sales"]
)

plt.xlabel("Advertising")
plt.ylabel("Sales")

plt.title("Advertising vs Sales")

plt.show()
```
